```python
import librosa
import numpy as np
import pandas as pd

def extract_hc_features(audio_path):
    speech, sr = librosa.load(audio_path, sr=None)
    
    features = []
    
    # MFCCs (13 coefficients - mean & std)
    mfccs = librosa.feature.mfcc(y=speech, sr=sr, n_mfcc=13)
    features.extend(np.mean(mfccs, axis=1))
    features.extend(np.std(mfccs, axis=1))
    
    # Delta MFCCs
    delta_mfccs = librosa.feature.delta(mfccs)
    features.extend(np.mean(delta_mfccs, axis=1))
    features.extend(np.std(delta_mfccs, axis=1))
    
    # Delta-Delta MFCCs
    delta2_mfccs = librosa.feature.delta(mfccs, order=2)
    features.extend(np.mean(delta2_mfccs, axis=1))
    features.extend(np.std(delta2_mfccs, axis=1))
    
    # Spectral features
    spec_centroid = librosa.feature.spectral_centroid(y=speech, sr=sr)
    features.append(np.mean(spec_centroid))
    features.append(np.std(spec_centroid))
    
    spec_bandwidth = librosa.feature.spectral_bandwidth(y=speech, sr=sr)
    features.append(np.mean(spec_bandwidth))
    features.append(np.std(spec_bandwidth))
    
    spec_rolloff = librosa.feature.spectral_rolloff(y=speech, sr=sr)
    features.append(np.mean(spec_rolloff))
    features.append(np.std(spec_rolloff))
    
    spec_contrast = librosa.feature.spectral_contrast(y=speech, sr=sr)
    features.extend(np.mean(spec_contrast, axis=1))
    features.extend(np.std(spec_contrast, axis=1))
    
    # Zero crossing rate
    zcr = librosa.feature.zero_crossing_rate(speech)
    features.append(np.mean(zcr))
    features.append(np.std(zcr))
    
    # RMS Energy
    rms = librosa.feature.rms(y=speech)
    features.append(np.mean(rms))
    features.append(np.std(rms))
    
    # Chroma
    chroma = librosa.feature.chroma_stft(y=speech, sr=sr)
    features.extend(np.mean(chroma, axis=1))
    features.extend(np.std(chroma, axis=1))
    
    return np.array(features)

# Extract features
my_audio_features = extract_hc_features('my_audio.wav')
print("Extracted features shape:", my_audio_features.shape)

# HCfeatures training data load karo
hc_train = pd.read_csv('../Feature Extraction/Features/HCfeatures.csv', header=None)
print("Training HCfeatures shape:", hc_train.shape)
print("Training features count (excluding filename):", hc_train.shape[1] - 1)
```

```python
def extract_hc_features(audio_path):
    speech, sr = librosa.load(audio_path, sr=None)
    
    features = []
    
    # MFCCs (13 coefficients - mean & std)
    mfccs = librosa.feature.mfcc(y=speech, sr=sr, n_mfcc=13)
    features.extend(np.mean(mfccs, axis=1))
    features.extend(np.std(mfccs, axis=1))
    
    # Delta MFCCs
    delta_mfccs = librosa.feature.delta(mfccs)
    features.extend(np.mean(delta_mfccs, axis=1))
    features.extend(np.std(delta_mfccs, axis=1))
    
    # Delta-Delta MFCCs
    delta2_mfccs = librosa.feature.delta(mfccs, order=2)
    features.extend(np.mean(delta2_mfccs, axis=1))
    features.extend(np.std(delta2_mfccs, axis=1))
    
    # Spectral features
    spec_centroid = librosa.feature.spectral_centroid(y=speech, sr=sr)
    features.append(np.mean(spec_centroid))
    features.append(np.std(spec_centroid))
    
    spec_bandwidth = librosa.feature.spectral_bandwidth(y=speech, sr=sr)
    features.append(np.mean(spec_bandwidth))
    features.append(np.std(spec_bandwidth))
    
    spec_rolloff = librosa.feature.spectral_rolloff(y=speech, sr=sr)
    features.append(np.mean(spec_rolloff))
    features.append(np.std(spec_rolloff))
    
    spec_contrast = librosa.feature.spectral_contrast(y=speech, sr=sr)
    features.extend(np.mean(spec_contrast, axis=1))
    features.extend(np.std(spec_contrast, axis=1))
    
    # Zero crossing rate
    zcr = librosa.feature.zero_crossing_rate(speech)
    features.append(np.mean(zcr))
    features.append(np.std(zcr))
    
    # RMS Energy
    rms = librosa.feature.rms(y=speech)
    features.append(np.mean(rms))
    features.append(np.std(rms))
    
    # Chroma
    chroma = librosa.feature.chroma_stft(y=speech, sr=sr)
    features.extend(np.mean(chroma, axis=1))
    features.extend(np.std(chroma, axis=1))
    
    # Tonnetz (6 features - mean & std = 12 more)
    tonnetz = librosa.feature.tonnetz(y=speech, sr=sr)
    features.extend(np.mean(tonnetz, axis=1))
    features.extend(np.std(tonnetz, axis=1))
    
    # Mel spectrogram (1 mean + 1 std = 2 more)
    mel = librosa.feature.melspectrogram(y=speech, sr=sr)
    features.append(np.mean(mel))
    features.append(np.std(mel))
    
    return np.array(features)

# Extract karo
my_audio_features = extract_hc_features('my_audio.wav')
print("Extracted features shape:", my_audio_features.shape)
```

```python
    # Labels load karo
labels = pd.read_csv('/Users/prateekchoudhary/Downloads/transfer_12029461_files_10a77b58/StressID Dataset/labels.csv', sep=",", header=0)
labels = labels.rename(columns={'subject/task': 'id'})

# HC training data load karo
hc_train = pd.read_csv('../Feature Extraction/Features/HCfeatures.csv', header=None)

# Pehla column filename hai, baaki 140 features hain
hc_train_ids = hc_train.iloc[:, 0]  # filenames
hc_train_features = hc_train.iloc[:, 1:]  # 140 features

# Id column clean karo (sirf subject/task part nikalo)
hc_train['id'] = hc_train_ids.apply(lambda x: x.replace('.wav', ''))

# Labels ke saath merge karo
hc_train_merged = hc_train.merge(labels, on='id', how='inner')
print("Merged shape:", hc_train_merged.shape)

# Train karo
label_col = 'binary-stress'
X_train = hc_train_merged.iloc[:, 1:141]
y_train = hc_train_merged[label_col]

from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

model_audio = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', MLPClassifier(max_iter=500, random_state=0, hidden_layer_sizes=(128, 128)))
])
model_audio.fit(X_train, y_train)

# Predict karo
my_audio_x = my_audio_features.reshape(1, -1)
prediction = model_audio.predict(my_audio_x)
proba = model_audio.predict_proba(my_audio_x)

label_map = {0: '😊 Relaxed / No Stress', 1: '😟 Stressed / Depressed'}
print("\n=== AUDIO MODEL RESULT ===")
print("Result:", label_map.get(prediction[0], prediction[0]))
print("Confidence:", dict(zip(model_audio.classes_, proba[0])))
```

```python
video_stress_prob = 0.9999999997  # video model se
audio_stress_prob = proba[0][1]   # audio model se

# Average combine
combined_stress_prob = (video_stress_prob + audio_stress_prob) / 2

print("=" * 40)
print("=== MULTIMODAL FINAL RESULT ===")
print("=" * 40)
print(f"🎥 Video Stress Probability:    {video_stress_prob:.4f}")
print(f"🎵 Audio Stress Probability:    {audio_stress_prob:.4f}")
print(f"🔗 Combined Stress Probability: {combined_stress_prob:.4f}")
print()
if combined_stress_prob > 0.5:
    print("🔴 Final Result: 😟 STRESSED / DEPRESSED")
else:
    print("🟢 Final Result: 😊 RELAXED / NO STRESS")
```
